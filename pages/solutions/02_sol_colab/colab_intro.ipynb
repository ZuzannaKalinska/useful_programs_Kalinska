### Obliczanie rozpływu prądów w obwodzie (Metoda Oczkowa)

Rozważmy prosty obwód elektryczny z dwoma oczkami, gdzie szukamy prądów $I_1$, $I_2$ oraz $I_3$ przy znanych rezystancjach i napięciu źródła.

Zastosujemy metodę układu równań liniowych wynikających z praw Kirchhoffa:

1. Napięciowe Prawo Kirchhoffa dla oczka 1:

   $$
   R_1 I_1 + R_3 I_3 = U.
   $$

2. Napięciowe Prawo Kirchhoffa dla oczka 2:

   $$
   R_2 I_2 - R_3 I_3 = 0.
   $$

3. Prądowe Prawo Kirchhoffa dla węzła:

   $$
   I_1 - I_2 - I_3 = 0.
   $$

```python
# @title Schemat obwodu
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 4))

# Rysowanie linii obwodu
ax.plot([1, 7], [3, 3], 'k-', lw=2)
ax.plot([1, 1], [1, 3], 'k-', lw=2)
ax.plot([7, 7], [1, 3], 'k-', lw=2)
ax.plot([1, 7], [1, 1], 'k-', lw=2)
ax.plot([4, 4], [1, 3], 'k-', lw=2)

# Źródło napięcia U
ax.plot([1, 1], [1.8, 2.2], 'k-', lw=4)
ax.plot([0.8, 1.2], [2.2, 2.2], 'k-', lw=2)
ax.plot([0.9, 1.1], [1.8, 1.8], 'k-', lw=2)
ax.text(0.5, 2.0, 'U', fontsize=12, fontweight='bold')

# Rezystor R1
rect1 = plt.Rectangle((2, 2.8), 1, 0.4, facecolor='white', edgecolor='black', lw=2)
ax.add_patch(rect1)
ax.text(2.3, 3.3, 'R1', fontsize=12, fontweight='bold')

# Rezystor R2
rect2 = plt.Rectangle((5, 2.8), 1, 0.4, facecolor='white', edgecolor='black', lw=2)
ax.add_patch(rect2)
ax.text(5.3, 3.3, 'R2', fontsize=12, fontweight='bold')

# Rezystor R3
rect3 = plt.Rectangle((3.8, 1.5), 0.4, 1, facecolor='white', edgecolor='black', lw=2)
ax.add_patch(rect3)
ax.text(4.4, 1.9, 'R3', fontsize=12, fontweight='bold')

# Strzałki prądów
ax.annotate('', xy=(2.8, 3.0), xytext=(1.8, 3.0), arrowprops=dict(arrowstyle="->", color='red', lw=2))
ax.text(2.1, 2.6, 'I1', color='red', fontsize=12, fontweight='bold')

ax.annotate('', xy=(6.2, 3.0), xytext=(5.2, 3.0), arrowprops=dict(arrowstyle="->", color='red', lw=2))
ax.text(5.5, 2.6, 'I2', color='red', fontsize=12, fontweight='bold')

ax.annotate('', xy=(4.0, 1.3), xytext=(4.0, 2.3), arrowprops=dict(arrowstyle="->", color='red', lw=2))
ax.text(3.6, 1.4, 'I3', color='red', fontsize=12, fontweight='bold')

ax.set_xlim(0, 8)
ax.set_ylim(0.5, 4)
ax.axis('off')
plt.title('Schemat obwodu dwuoczkowego', fontsize=14, fontweight='bold')
plt.show()
```

### Wykres zależności $y = k^2$

Poniższy wykres przedstawia zależność kwadratową dla kolejnych liczb całkowitych $k$ w zakresie od $1$ do $20$.

Każdy punkt na wykresie odpowiada parze współrzędnych $(k, k^2)$, co pozwala zobrazować, jak szybko rośnie funkcja kwadratowa w tym przedziale. Wykres został automatycznie zapisany do pliku `wykres.png`.

```python
import matplotlib.pyplot as plt

# Przygotowanie danych
k = list(range(1, 21))
y = [val**2 for val in k]

# Tworzenie wykresu
plt.figure(figsize=(10, 6))
plt.scatter(k, y, color='b', label='$y = k^2$', zorder=3)
plt.title(r'Wykres funkcji $y = k^2$ dla $k = 1, \dots, 20$', fontsize=14)
plt.xlabel('k', fontsize=12)
plt.ylabel('y', fontsize=12)
plt.xticks(k)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()

# Zapisanie wykresu do pliku wykres.png
plt.savefig('wykres.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML
import base64
import warnings

# Ignorowanie ostrzeżeń dla czystego wyjścia
warnings.filterwarnings('ignore')

# Parametry wahadła
g = 9.81  # przyspieszenie ziemskie [m/s^2]
L = 1.0   # długość wahadła [m]
theta0 = np.radians(45)  # maksymalne wychylenie początkowe (45 stopni)

# Czas animacji
t_max = 5.0
fps = 30
dt = 1.0 / fps
t = np.arange(0, t_max, dt)

# Uproszczone rozwiązanie analityczne (dla małych kątów, ale dobrze wygląda w animacji)
omega = np.sqrt(g / L)
theta = theta0 * np.cos(omega * t)

# Współrzędne x, y końca wahadła
x = L * np.sin(theta)
y = -L * np.cos(theta)

# Przygotowanie wykresu i obiektów do rysowania (zmniejszamy wysokość figury i dopasowujemy ylim)
fig, ax = plt.subplots(figsize=(6, 4.2))
ax.set_xlim(-1.2 * L, 1.2 * L)
ax.set_ylim(-1.05 * L, 0.1 * L)           # Przycięte limity na osi Y
ax.set_aspect('equal')
ax.grid(True, linestyle='--', alpha=0.5)
ax.set_title("Animacja wahadła matematycznego", fontsize=12)

# Ręczne dopasowanie marginesów, aby uniknąć konieczności stosowania bbox_inches='tight'
plt.subplots_adjust(left=0.1, right=0.9, top=0.88, bottom=0.1)

# Elementy graficzne
line, = ax.plot([], [], 'o-', lw=2, color='black', markersize=6)  # linka i zaczep
bob, = ax.plot([], [], 'o', color='red', markersize=15)            # masa (kulka)

def init():
    line.set_data([], [])
    bob.set_data([], [])
    return line, bob

def update(frame):
    # Współrzędne punktu zawieszenia to (0,0)
    this_x = [0, x[frame]]
    this_y = [0, y[frame]]
    line.set_data(this_x, this_y)
    bob.set_data([x[frame]], [y[frame]])
    return line, bob

# Utworzenie animacji
ani = animation.FuncAnimation(
    fig, update, frames=len(t), init_func=init, blit=True, interval=1000/fps
)

# Zapisanie animacji do pliku video.mp4 bez ostrzeżeń
ani.save('video.mp4', writer='ffmpeg', fps=fps, dpi=150)
plt.close(fig)

# Kodowanie wideo w formacie Base64 do wyświetlenia w HTML
video_file = open('video.mp4', 'rb').read()
video_b64 = base64.b64encode(video_file).decode('ascii')

# Wyświetlenie odtwarzacza wideo w notatniku
HTML(f'''
<video width="400" height="280" controls>
    <source src="data:video/mp4;base64,{video_b64}" type="video/mp4">
    Twoja przeglądarka nie obsługuje tagu wideo.
</video>
''')
```

```python
# Modyfikacja funkcji parametrycznych i generowanie nowej krzywej trójwymiarowej
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Parametry nowej krzywej (np. trójwymiarowa spirala stożkowa)
t = np.linspace(0, 10 * np.pi, 1000)
x = t * np.sin(t)
y = t * np.cos(t)
z = t

# Tworzenie wykresu 3D
fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')
ax.plot(x, y, z, label=r'Spirala stożkowa: $x=t\sin(t), y=t\cos(t), z=t$', color='magenta', lw=2)

# Ustawienia wykresu
ax.set_title("Zmodowana 3D Krzywa Parametryczna", fontsize=14, fontweight='bold')
ax.set_xlabel("x(t)", fontsize=11)
ax.set_ylabel("y(t)", fontsize=11)
ax.set_zlabel("z(t)", fontsize=11)
ax.legend(loc='upper left')
ax.grid(True)

# Zapisanie wyniku do pliku parametric_curve.png
plt.savefig('parametric_curve.png', dpi=300, bbox_inches='tight')
plt.show()
```
